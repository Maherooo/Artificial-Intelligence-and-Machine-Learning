Which team won the most mathches?

```python
match_wins=df['match_winner'].value_counts()
match_wins
```

```python
sns.barplot(x=match_wins.index,y=match_wins.values,palette='viridis')
plt.title("Most Match win by team")
plt.xticks(rotation=45, ha='right') 
plt.tight_layout()
```

<h2>Toss Dicision Trends</h2>

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title("Toss Decision Trends")
```

<h3>Is Toss winner Wins the match</h3>

```python
count=df[df['toss_winner']==df['match_winner']]['match_id'].count()
parcentage=(count*100)/df.shape[0]
print(f"Toss win and Match Win Percentage: {round(parcentage, 2)} %")
```

<h1>Top Scorers</h1>

```python
df.head()
```

```python
high=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

```python
df['highest_wickets'] =df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets']=df['highest_wickets'].astype(int)
top_bowler=df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowler.plot(kind='barh')
```
