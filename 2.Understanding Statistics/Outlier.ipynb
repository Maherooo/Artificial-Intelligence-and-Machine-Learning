```python
import numpy as np
```

```python
arr=np.array([2,3,4,5,6,8,9,11,14,15,17,19,202])
arr
```

```python
Q1=np.percentile(arr,25)
Q3=np.percentile(arr,75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
UF=Q3+1.5*(IQR)
UF
```

```python
LF=Q1-(1.5*IQR)
LF
```

```python
l=[]
for i in arr:
    if i>=LF and i<=UF:
        l.append(i);
l
```

```python
arr2=np.array(l)
print(arr2)
```
